```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
os.chdir("..") if 'notebooks' in os.getcwd() else None

df = pd.read_csv('data/raw/soil_india_shc.csv')
df.head()
corr_matrix = df[['pH', 'EC', 'OC', 'N', 'P', 'K', 'Moisture']].corr()
corr_matrix
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix of Soil Features")
plt.show()
corr_matrix.to_csv('data/processed/soil_correlation_matrix.csv', index=True)

# Save correlation matrix for other members
corr_matrix = df.corr(numeric_only=True)
corr_matrix.to_csv('data/processed/soil_correlation_matrix.csv', index=True)
print("✅ Correlation matrix saved to 'data/processed/soil_correlation_matrix.csv'")


```
