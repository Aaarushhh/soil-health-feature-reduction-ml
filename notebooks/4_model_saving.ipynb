```python
import joblib
from sklearn.ensemble import RandomForestClassifier
import pandas as pd

# Load processed dataset
df = pd.read_csv('../data/processed/soil_reduced_features.csv')

# Encoding (same as Member 3)
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
for col in ['State', 'District', 'Farmer_ID', 'Soil_Type', 'Crop_Suitability']:
    df[col] = le.fit_transform(df[col])

# Split into features and target
X = df.drop(columns=['Crop_Suitability'])
y = df['Crop_Suitability']

# Train Random Forest
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X, y)
import os

# Create models directory if it doesn't exist
os.makedirs('../models', exist_ok=True)

# Save model
joblib.dump(rf, '../models/random_forest_model.joblib')
print("✅ Model saved successfully at '../models/random_forest_model.joblib'")

# Save model
joblib.dump(rf, '../models/random_forest_model.joblib')
print("✅ Model saved successfully at '../models/random_forest_model.joblib'")
```
