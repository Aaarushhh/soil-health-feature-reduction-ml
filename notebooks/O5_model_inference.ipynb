```python
import pandas as pd
import joblib
from sklearn.preprocessing import LabelEncoder

# Load trained model
rf = joblib.load('../models/random_forest_model.joblib')
print("✅ Model loaded successfully!")

# Load dataset
df = pd.read_csv('../data/processed/soil_reduced_features.csv')

# Encode categorical columns (same as training step)
categorical_cols = ['State', 'District', 'Farmer_ID', 'Soil_Type']

le_dict = {}
for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    le_dict[col] = le  # store for decoding later if needed

print("🔢 Categorical columns encoded!")

# Select a sample for prediction
sample = df.iloc[0:1].drop(columns=['Crop_Suitability'], errors='ignore')

# Make prediction
prediction = rf.predict(sample)
print("🌱 Predicted Crop Suitability Class (encoded):", prediction[0])

# Decode prediction if target column exists
if 'Crop_Suitability' in df.columns:
    target_le = LabelEncoder()
    df['Crop_Suitability'] = target_le.fit_transform(df['Crop_Suitability'])
    decoded_label = target_le.inverse_transform([prediction[0]])
    print("🌾 Predicted Crop Suitability:", decoded_label[0])
else:
    print("🌾 Predicted Crop Suitability (decoded): [target label encoder not available]")
```
